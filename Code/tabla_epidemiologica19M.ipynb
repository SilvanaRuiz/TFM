```python
from scipy.stats import mannwhitneyu

def mannwhitney(df,variable_continua, variable_grupo):

    grupo_0 = df[df[variable_grupo] == 0][variable_continua].dropna()
    grupo_1 = df[df[variable_grupo] == 1][variable_continua].dropna()

    stat, p = mannwhitneyu(grupo_0, grupo_1, alternative='two-sided')

    print(f"\nMann-Whitney U test para '{variable_continua}' entre grupos en '{variable_grupo}'")
    print(f"Mediana grupo 0 (control): {grupo_0.median()}")
    print(f"Mediana grupo 1 (caso): {grupo_1.median()}")
    print(f"U = {stat:.4f}, P-valor = {p:.4f}")

def mannwhitney_por_sexo(df,  variable_continua, variable_grupo,):
   

    for sexo in ['M', 'V']:  # Ajusta si usas otros códigos
        sub_df = df[df['SEXO'] == sexo]
        print(f"\n--- Análisis en {'Mujeres' if sexo == 'M' else 'Hombres'} ---")

        grupo_0 = sub_df[sub_df[variable_grupo] == 0][variable_continua].dropna()
        grupo_1 = sub_df[sub_df[variable_grupo] == 1][variable_continua].dropna()

        if len(grupo_0) < 2 or len(grupo_1) < 2:
            print("No hay suficientes datos para aplicar la prueba.\n")
            continue

        stat, p = mannwhitneyu(grupo_0, grupo_1, alternative='two-sided')

        print(f"Mann-Whitney U test entre grupos para '{variable_continua}' en {variable_grupo}")
        print(f"Mediana grupo 0 (control): {grupo_0.median()}")
        print(f"Mediana grupo 1 (caso): {grupo_1.median()}")
        print(f"U = {stat:.4f}, P-valor = {p:.4f}")
```

```python
from scipy.stats import chi2_contingency
from statsmodels.stats.proportion import proportions_ztest
import pandas as pd
from scipy.stats import chi2_contingency
from statsmodels.stats.proportion import proportions_ztest
import pandas as pd

def chi_cuadrado(df, variable_grupo, variable_binaria):
    categorias = df[variable_grupo].dropna().unique()

    if len(categorias) > 1:
        try:
            tabla = pd.crosstab(df[variable_grupo], df[variable_binaria])

            chi2, p, dof, expected = chi2_contingency(tabla)

            # 💡 Chequeo de frecuencias esperadas
            if (expected < 5).any():
                print("⚠️ Algunas frecuencias esperadas son menores a 5. Considera usar Fisher’s Exact Test si la tabla es 2x2.")

            print(f"\n🔎 Chi-cuadrado entre '{variable_grupo}' y '{variable_binaria}'")
            print("Tabla de contingencia:")
            print(tabla)
            print(f"\nChi² = {chi2:.4f}")
            print(f"P-valor = {p:.4f}")
            print(f"Grados de libertad = {dof}")
            print("Frecuencias esperadas:")
            print(pd.DataFrame(expected, index=tabla.index, columns=tabla.columns))

        except Exception as e:
            print(f"❌ Error en Chi-cuadrado: {e}")

    else:
        print(f"\n📌 Solo se encontró una categoría en '{variable_grupo}': '{categorias[0]}'.")
        try:
            sub_df = df[df[variable_grupo] == categorias[0]]
            tabla = pd.crosstab(sub_df[variable_grupo], sub_df[variable_binaria])

            if tabla.shape[1] == 2 and tabla.shape[0] == 1:
                # 💡 Realizar Test de Proporciones si tabla es 1x2
                count_success = tabla.iloc[0,1]  # Número de éxitos
                count_failure = tabla.iloc[0,0]  # Número de fracasos

                successes = [count_failure, count_success]
                nobs = [count_failure + count_success, count_failure + count_success]

                print("\n✅ Realizando Test de Proporciones (proportions_ztest)...")
                stat, p_value = proportions_ztest(successes, nobs, alternative='two-sided')
                print(f"Éxitos (1): {count_success} | Fracasos (0): {count_failure} | Total: {nobs[0]}")
                print(f"Z = {stat:.4f} | P-valor = {p_value:.4f}")

            else:
                print("⚠️ No hay dos columnas en la variable binaria o más de una fila. No se puede realizar Test de Proporciones.\n")

        except Exception as e:
            print(f"❌ Error en test de proporciones: {e}")
def chi_cuadrado_por_sexo(df, variable_grupo, variable_binaria):
    for sexo in ['V', 'M']:  
        sub_df = df[df['SEXO'] == sexo]
        print(f"\n--- Análisis en {'Mujeres' if sexo == 'M' else 'Hombres'} ---")

        categorias = sub_df[variable_grupo].dropna().unique()

        if len(categorias) > 1:
            try:
                tabla = pd.crosstab(sub_df[variable_grupo], sub_df[variable_binaria])

                chi2, p, dof, expected = chi2_contingency(tabla)

                if (expected < 5).any():
                    print("⚠️ Algunas frecuencias esperadas son menores a 5. Considera usar Fisher’s Exact Test si la tabla es 2x2.")

                print(f"Chi-cuadrado entre '{variable_grupo}' y '{variable_binaria}'")
                print("Tabla de contingencia:")
                print(tabla)
                print(f"\nChi² = {chi2:.4f}")
                print(f"P-valor = {p:.4f}")
                print(f"Grados de libertad = {dof}")
                print("Frecuencias esperadas:")
                print(pd.DataFrame(expected, index=tabla.index, columns=tabla.columns))

            except Exception as e:
                print(f"❌ Error al analizar {sexo}: {e}")

        else:
            # Solo una categoría presente
            print(f"\n📌 Solo se encontró una categoría en '{variable_grupo}': '{categorias[0]}'.")
            try:
                sub_sub_df = sub_df[sub_df[variable_grupo] == categorias[0]]
                tabla = pd.crosstab(sub_sub_df[variable_grupo], sub_sub_df[variable_binaria])

                if tabla.shape[1] == 2 and tabla.shape[0] == 1:
                    # 💡 Hacer test de proporciones (proportions_ztest) si tabla es 1x2
                    count_success = tabla.iloc[0,1]  # Número de éxitos (por ejemplo, autolesiones sí)
                    count_failure = tabla.iloc[0,0] # Número de fracasos
                    successes = [count_failure, count_success]
                    nobs = [count_failure + count_success, count_failure + count_success]

                    print("\n✅ Realizando Test de Proporciones (proportions_ztest)...")
                    stat, p_value = proportions_ztest(successes, nobs, alternative='two-sided')
                    print(f"Éxitos (1): {count_success} | Fracasos (0): {count_failure} | Total: {nobs[0]}")
                    print(f"Z = {stat:.4f} | P-valor = {p_value:.4f}")

                else:
                    print("⚠️ No hay dos columnas en la variable binaria o más de una fila. No se puede realizar Test de Proporciones.\n")

            except Exception as e:
                print(f"❌ Error en categoría única para {sexo}: {e}")
```
